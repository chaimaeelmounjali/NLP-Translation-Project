# Projet NLP - Text Mining : Traduction Automatique Darija

Ce dépôt contient le pipeline de traitement, de nettoyage et de validation d'un dataset parallèle pour la traduction automatique impliquant la Darija marocaine, l'Arabe Standard (MSA) et l'Anglais.

L'objectif de ce projet est de transformer un corpus brut de données pré-alignées (un "Silver Standard" de 10 000 lignes) en un **"Gold Standard"** propre et fiable, qui pourra servir de vérité terrain (Ground Truth) et de base d'entraînement pour un modèle performant (AraT5v2).

## 📌 Approche et Méthodologie

Notre pipeline est structuré en **trois grandes phases** complémentaires :

### Phase 1 : Exploration (EDA) et Nettoyage
Nous partons de données initiales contenant potentiellement du bruit et des artefacts d'IA (balises de tokenisation, erreurs de script, URLs, ponctuations superflues).
- **Audit quantitatif et qualitatif** : Vérification des doublons (0 trouvé), analyse de la longueur des textes et détection d'anomalies de format (ex: arabe dans la colonne arabizi).
- **Nettoyage Automatique Python** : Utilisation d'expressions régulières (`regex`) pour supprimer les URLs, emojis, les tokens `<unk>`, et normaliser la ponctuation. Les données sont ensuite séparées en deux corpus distincts (Silver Shard et Gold Shard).

### Phase 2 : Stratégie de Validation Structurée
Selon la nature du fragment de données, nous appliquons deux types de validation :
- **Validation Humaine (Gold Set)** : 1 000 échantillons sont extraits et chargés dans l'outil **Label Studio**. L'humain corrige manuellement la vérité terrain, en appliquant des règles linguistiques strictes (système de racines sémitiques, articles définis, transcriptions phonétiques exactes de l'Arabizi avec les chiffres 3, 7, 9, etc.).
- **Auto-Correction IA (Silver Set)** : 9 000 échantillons sont traités en masse via une API LLM (GPT-4) pour corriger et normaliser les lignes (en résolvant notamment les fuites de scripts).

### Phase 3 : Analyse des Erreurs et Bilan
Nous comparons les fichiers à différentes étapes (Avant IA vs Après IA, Avant Validation Manuelle vs Après) pour chiffrer l'impact de nos corrections et l'évolution de la qualité.
